```python
#real data
# Movies Data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

plt.plot(years, oscar_revenue)
plt.title("Oscar Years vs Revenue(in $M)")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")
```

```python
# Plot multiple
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(years, oscar_revenue, label="Oscar Winners")
plt.plot(non_oscar_years, non_oscar_revenue, label="Non-Oscar Winners")
plt.title("Years vs Revenue(in $M)")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")
plt.legend()
# plt.legend(loc="upper right")
# plt.legend(loc="upper left")
# plt.legend(loc="center")
```

```python
# using Format Strings
plt.plot(years, oscar_revenue, "o-b", label="Oscar Winners")
plt.plot(non_oscar_years, non_oscar_revenue, ">--g", label="Non-Oscar Winners")

plt.plot(years, oscar_revenue, color="black", linestyle="-", marker="o",label="Oscar Winners")
plt.plot(non_oscar_years, non_oscar_revenue, color="red", linestyle="--", marker=">", linewidth=5, label="Non-Oscar Winners")
plt.grid()
```

```python
# Bar charts
movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]  # in $M

plt.bar(years, revenue, color="purple")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.title("Yearly revenue of movies")

# Add labels on top of bars
for i in range(len(revenue)):
    plt.text(years[i], revenue[i] + 20, str(revenue[i]), ha="center")  
    # text(pos of text acc to x-axis, pos of y, text, alignment)

plt.ylim(0, max(revenue)+200) #lower, upper limit - add space
```

```python
# comparing data with bar charts
oscar_movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

years = [2008, 2009, 2010, 2011, 2012]

# Plotting Bar Chart
import numpy as np
x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Oscar")
plt.bar(x+width/2, non_oscar_revenue, width, label="Non-Oscar")

plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.title("Yearly revenue of Oscar vs Non-Oscar movies")
plt.legend()
plt.tight_layout()

plt.xticks(x, years) # to show x-axis labels
```

```python
# Horizontal Bar Chart
oscar_movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

years = [2008, 2009, 2010, 2011, 2012]

# Plotting
# y = np.arange(len(years))
plt.barh(oscar_movies, oscar_revenue)
plt.xlabel("Revenue")
plt.ylabel("Movie")
plt.title("Revenue for each Movie")
```

```python
# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

# plot
# plt.scatter(age, blood_pressure, color="red")

plt.title("Age vs Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()

# Customizing plots
colors = ["green" if bp<135 else "red" for bp in blood_pressure]
sizes = blood_pressure
plt.scatter(age, blood_pressure, color=colors, s=sizes, marker='^', alpha=0.5)

# plt.scatter(age, blood_pressure, c=blood_pressure, cmap="plasma")
plt.scatter(age, blood_pressure, c=blood_pressure, cmap="OrRd")# makes more sense
# plt.scatter(age, blood_pressure, c=[5]*len(age), cmap="plasma")
plt.colorbar(label="BP")

for i in range(len(age)):
    plt.annotate(people[i], xy=(age[i], blood_pressure[i]))
plt.xlim(min(age), max(age)+10)
```

```python
 #Multiple scatter plots
cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

# Plots
plt.scatter(winter_temp, winter_humidity, color="blue", label="Winter")
plt.scatter(summer_temp, summer_humidity, color="orange", label="Summer")

plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.legend()
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels=expenses)

# set style
plt.style.use("ggplot")

# add %-age of each slice
plt.pie(amounts, labels=expenses,  autopct='%1.1f%%')

#set colors
colors = ["purple", "green", "blue", "pink", "yellow"]
plt.pie(amounts, labels=expenses,  autopct='%1.1f%%', colors=colors)

# wedgeprops
plt.pie(amounts, labels=expenses,  autopct='%1.1f%%', 
        wedgeprops={"edgecolor":"black", "linewidth": 1, "linestyle":"-.", "fill":False}
       )
```

```python
 #explode
explode = [0, 0, 0, 0.2, 0]
# plt.pie(amounts, labels=expenses, explode=explode)

# shadow
# plt.pie(amounts, labels=expenses, explode=explode, shadow=True)

# angle
plt.pie(amounts, labels=expenses, explode=explode, shadow=True, startangle=90) # top
plt.pie(amounts, labels=expenses, explode=explode, shadow=True, startangle=180) # flips
```

```python
# Histograms

# generate random data
import numpy as np

np.random.seed(0)
# generates 100 nums with mean 70 & std dev 10
scores = np.random.normal(loc=70, scale=10, size=100) 


plt.hist(scores, bins=5, color="skyblue", edgecolor="black")
plt.title("Distribution of exam scores")
plt.xlabel("Marks")
plt.ylabel("Number of students")
```

```python
# custom bins
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

# custom bins
bins = [0, 30, 70, 90, 100]
plt.hist(
    marks,
    bins=bins,
    edgecolor="black",
    color="lightblue"
)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks of 30 Students")
plt.grid(True)
```

```python
# axv line
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

# custom bins
bins = [0, 30, 70, 90, 100]
plt.hist(
    marks,
    bins=bins,
    edgecolor="black",
    color="lightblue"
)

# line at passing marks - 33
plt.axvline(
    x=33,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Pass Mark (33)"
)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks of 30 Students")
plt.grid(True)
```

```python
# Multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.hist(fraud_transactions, bins=10, label='Fraud', color='red', alpha=0.5, edgecolor="black")
plt.hist(legit_transactions, bins=10, label='Legit', color='green', alpha=0.5, edgecolor="black")

plt.xlabel('Transaction Amount ($)')
plt.ylabel('Frequency')
plt.title('Transaction Amount Distribution: Legitimate vs Fraudulent')
plt.legend()
plt.grid(True)
```
